# How to choose the indexing parameters

`python -m anri.index` runs in six stages (histogram, search, filter, fit, localsearch, populations), each of which writes one file that the next one reads. This notebook runs the same stages on your dataset, with the plots that help you choose each stage's options before you run it, in the spirit of ImageD11's S3DXRD notebooks (`tomo_1_index`, `tomo_2_map`, `pbp_1_indexing`, `pbp_3_refinement`).

| Section | Stage | Options | What the plots show |
|---|---|---|---|
| 1. Rings | histogram | `--rings`, `--ring-merge`, `--tth-tol`, `--cif` | the 2θ profile and the (2θ, η) cake: which rings are clean, how wide they are, what lies between them; pick any rings, not only the first |
| 2. Data | histogram | `--monitor`, `--gridstep`, `--detector-mask` | intensity per dty row and per ω: flux steps between the rows' scans; the share of each ring the detector measures |
| 3. Rotation axis, mask | fit | `--y0`, `--mask` | the sinogram's centre, a back-projection for any y0, and the sample mask from it (thresholded or drawn) |
| 4. Lit map | search | `--lit`, `--etacut` | how much of the map is lit against how much intensity it holds; what the η cut removes |
| 5. Grid | search | `--grid`, `--max-chance` | chance completeness and size of every grid step, and how much each finer step improves the fit |
| 6. Pruning | filter | `--prune`, `--min-comp`, `--min-lr`, `--keep` | completeness and likelihood ratio of every orientation, and how many each cut keeps |
| 7. Occupancy | fit | `--cand`, `--iter`, `--coarse`, `--block-gb`, `--beam`, `--censor` | the deviance per iteration, how many voxels still change orientation, and whether the candidates suffice |
| 8. Model against data | | | what the fit leaves unexplained, per row, per ring and spot by spot |
| 9. Local search | localsearch | `--levels`, `--cover`, `--per-voxel` | how many orientations each voxel needs before it runs; after it, the deviance at each level and how far each voxel's main population moved |
| 10. Populations | populations | `--occupied`, `--min-frac` | the occupied voxels within the mask, and the populations per voxel |

Where a choice is cheap to redo, a slider redraws the plot, and **the value on the slider is what the later cells use**. The numbers come from `anri.index.diagnostics` and the plots from `anri.index.plots`; the stages are the command line's own functions, so the files written here are the ones `python -m anri.index` writes, with the same provenance (`python -m anri.index show <file>`). The last cell prints the command that remakes them.

Run it top to bottom; after moving a slider, re-run the cells below it. Figures are interactive (ipympl, `%matplotlib widget`): zoom and pan with each figure's toolbar, and the zoom stays while the sliders move. It needs `ipywidgets`, `ipympl` and `matplotlib`. Sections 7 to 10 run the full occupancy fit, which costs as much here as on the command line (set `run_fit = False` to stop after pruning on big datasets). For the figures to survive a restart of the Jupyter server, turn on JupyterLab's *Save Widget State Automatically* ([how](index.rst#figures-that-survive-a-restart)). `interactive = False` draws each figure once, at the parameters' values, as an inline image instead, as for these docs; the sliders then stay at their starting values, which the later cells use.

With no dataset given, it renders the 316L [phantom](../tutorials/phantom.ipynb) as in the [indexing tutorial](../tutorials/indexing.ipynb) (about a minute on a laptop CPU), so it runs out of the box.

```python
# Parameters: the options of python -m anri.index, at its defaults. The sliders below start from these values.

# the dataset, as python -m anri.index <analysisroot> <sample> <dataset>; None renders a phantom to try things on
analysisroot = None
sample = None
dataset = None
phase = None  # --phase: name in pars.json (None: the only one)
parfile = None  # --parfile (None: the DataSet's, else pars/pars.json beside PROCESSED_DATA)
cif = None  # --cif: CIF of the phase, for structure factors (None: |F|^2 = 1)
monitor = None  # --monitor: counter to normalise intensities by, e.g. "fpico6" (None: none)
dty_source = "dataset"  # --dty-source: which dty bins the frames into rows: "dataset" (each row's) or "frames" (each frame's own)
frame_dty = True  # follow dty's motion within the rows (fly and helical scans) in the model (--no-frame-dty to turn it off)
gridstep = 1  # --gridstep: voxel = gridstep x dty step, rows summed to match
grid_pad = 0.0  # --grid-pad: widen the voxel grid on every side (sample units), for a sample reaching past the scan
saturation = None  # --saturation: detector counts at which a pixel saturated, e.g. 65535 (None: none)
detector_mask = None  # --detector-mask: detector mask, non-zero for masked pixels (None: the DataSet's maskfile)
pol_factor = 1.0  # --pol-factor: degree of horizontal polarisation of the beam

# 1. rings
rings = 6  # --rings: the number of first rings, or which rings as indices from the lowest 2theta, e.g. [0, 1, 2, 4, 7]
ring_merge = 1e-4  # --ring-merge: reflections within this in d* (1 / angstrom) are one ring
tth_tol = None  # --tth-tol, deg (None: measured per ring)
# 3. rotation axis and sample mask
y0 = None  # --y0 (None: the DataSet's)
mask = "disc"  # --mask: "disc" (the voxels seen at every omega), "auto" (the sample's support), "threshold" (section 3), or a file
pad = 0.0  # --pad: grow the disc by this (sample units)
absorption = True  # with a CIF (or mu): model absorption through the mask (--no-absorption to turn it off)
mu = None  # --mu: attenuation coefficient, 1 / sample units (None: from the CIF)
# 4. lit map
lit = 1.0  # --lit: x the median non-empty bin
etacut = 0.2  # --etacut
# 5. grid
grid = None  # --grid, deg (None: the coarsest with chance completeness <= max_chance)
max_chance = 0.5  # --max-chance
# 6. pruning
prune = "likelihood"  # --prune: "likelihood" or "completeness"
min_comp = None  # --min-comp (None: the chance level for likelihood, halfway to the maximum for completeness)
min_lr = 25.0  # --min-lr
keep = 100000  # --keep
# 7. occupancy
n_cand = 64  # --cand
n_iter = 10  # --iter: SQUAREM iterations (3 MLEM updates each)
coarse = 1  # --coarse
block_gb = 1.0  # --block-gb
beam = 0.0  # --beam: FWHM of the beam across dty (0: the 2-row model, close to FWHM = dty step)
censor = 0.0  # --censor: counts per bin below which an empty bin agrees with the model (0: off)
# 9. local search
levels = 2  # --levels: each a third of the step before (0: no local search)
cover = 0.95  # --cover
per_voxel = 8  # --per-voxel
# 10. populations
occupied = 0.2  # --occupied
min_frac = 0.1  # --min-frac

n_cpu = 4  # --n-cpu
outdir = "."  # --outdir: where the stage files go

# this notebook only
interactive = True  # sliders (ipympl); False: static figures at the parameters' values, which survive a
# restart of the Jupyter server and show in the docs
previous = None  # a stage file (e.g. an earlier ..._populations.h5): its options, and those of the stages before it, replace the ones above
max_rings_shown = 10  # rings drawn on the 2theta profile, to choose the rings from (more if rings asks for more)
run_fit = True  # False: stop after pruning (sections 7-10 run the full occupancy fit)
draw = False  # draw the sample mask by hand in section 3 (needs interactive = True)
grid_scan = True  # fit the orientations of every grid step in section 5, to compare their deviance
```

```python
import anri.utils

anri.utils.setup(n_cpu=n_cpu)  # before any JAX computation

import os
import tempfile
import time
from types import SimpleNamespace

import h5py
import ipywidgets as widgets
import numpy as np
from IPython import get_ipython
from IPython.display import display
from matplotlib import pyplot as plt

import anri.index as ix
import anri.io
import anri.simulate
from anri.index import diagnostics as dg
from anri.index import plots as pl

get_ipython().run_line_magic("matplotlib", "widget" if interactive else "inline")
T0 = time.time()
NOTEBOOK = "indexing_parameters.ipynb"  # recorded as the command in each stage file's provenance


def log(msg):
    print(f"[{time.time() - T0:7.1f} s] {msg}", flush=True)


def figure(name, **kwargs):
    # a figure shown where display() puts it (under its sliders), not at the end of the cell; re-running the cell
    # replaces it
    plt.close(name)
    with plt.ioff():
        fig = plt.figure(name, layout="constrained", **kwargs)
    if interactive:
        fig.canvas.header_visible = False
    return fig


def show(fig):
    display(fig.canvas if interactive else fig)


def interact(fig, update, **controls):
    # sliders that call a plot's update, which redraws it and returns a summary; their values are in .kwargs
    if not interactive:  # draw once at the controls' starting values
        values = {k: c.value for k, c in controls.items()}
        print(update(**values))
        display(fig)
        return SimpleNamespace(kwargs=values)
    w = widgets.interactive(lambda **kw: print(update(**kw)), **controls)
    for a in fig.axes:  # fix the limits as first drawn, so that the sliders do not undo a zoom
        if getattr(a, "_colorbar", None) is None:
            a.set_xlim(a.get_xlim())
            a.set_ylim(a.get_ylim())
    display(w, fig.canvas)
    return w


def slider(value, lo, hi, step, name, log_scale=False):
    kind = widgets.FloatLogSlider if log_scale else widgets.FloatSlider
    extra = {"base": 10} if log_scale else {"readout_format": ".3g"}
    return kind(value=value, min=lo, max=hi, step=step, description=name, continuous_update=False, **extra)
```

## Remaking earlier files

To remake the files of an earlier run (or start from its choices), set `previous` in the parameters to one of its stage files, e.g. `..._populations.h5`. The cell below takes every option recorded in it and in the files before it (as `python -m anri.index show` lists them), the dataset included, in place of the parameters, and logs each one it changes. The sliders then start from them, so running top to bottom remakes the files. Options left to be worked out from the data (e.g. `tth_tol = None`) are not recorded, so keep them at their defaults in the parameters. The files are written to `outdir`, which must not be the folder `previous` is in.

```python
if previous is not None:
    stages = [anri.io.provenance.read_provenance(p) for p in reversed(anri.io.provenance.chain(previous))]
    recorded = {}
    for prov in stages:  # the first stage first: a later stage's value wins (censor, in filter and fit)
        recorded.update(prov["options"])
    recorded.pop("save_model", None)
    if "fit" in [s["stage"] for s in stages] and "localsearch" not in [s["stage"] for s in stages]:
        recorded["levels"] = 0
    names = {"cand": "n_cand"}  # the notebook's names, where they differ
    for k, v in recorded.items():
        if globals().get(names.get(k, k)) != v:
            log(f"{names.get(k, k)} = {v!r} (was {globals().get(names.get(k, k))!r})")
        globals()[names.get(k, k)] = v
    if os.path.samefile(outdir, os.path.dirname(os.path.abspath(previous))):
        raise ValueError(f"outdir is where {previous} is: the new files would overwrite it and the files before it; "
                         "set outdir to another folder")
```

## 1. Rings: `--rings`, `--ring-merge`, `--tth-tol`, `--cif`

Pixels count only within each ring's 2θ window, so the windows decide what the indexer sees. As the histogram stage does, the profile is measured on 9 scans spread over the dty range, and each ring's offset and half-width (holding 95% of its intensity above the local background) are measured from it; the window is |offset| + half-width, unless `tth_tol` is set (0 on the slider: measured per ring). Each ring's width is measured between its true neighbours, used or not, so a window never reaches into a ring that is left out.

**Which rings.** Choose them in the list (ctrl- or shift-click for several): any rings, not only the first. The table below the plots lists every ring shown, with its multiplicity, mean |F|² and the gap to the next ring, so pick strong rings well apart from their neighbours. The first rings are usually best for cubic metals; for low symmetry they can be weak, sit on low-angle scattering or crowd together, and rings further out, chosen one by one, do better. The selection is what the histogram stage uses, as `--rings 9,10,11` on the command line (a plain count when it is the first rings).

To look at the rings in another program (or fit the lattice by Rietveld), `python -m anri.index powder <analysisroot> <sample> <dataset>` (`anri.index.stages.stage_powder`) exports the whole scan as a powder pattern, `.h5` and `.xye`, each bin divided by the solid angle of the measured pixels in it.

**Merged rings.** Reflections closer than `ring_merge` in d* are one ring (one dashed line). Rings closer than the data can separate should be merged, not split between two windows that each take half the intensity: raise `ring_merge` (and re-run this cell) if two dashed lines sit within one peak of the profile.

What to look for:

- **Rings that overlap or touch** (the gap column, or windows that meet in the plots): their spots cannot be told apart, so a reflection is predicted in both. Leave one of them out, merge them (`ring_merge`), or use a tolerance below the gap.
- **Intensity between rings, or a ring that is not this phase's**: another phase, or a wrong lattice. It is ignored by the indexer, but if it falls in a window it lights the map for nothing.
- **Detector distortion widens every ring**: a pixel or two of distortion is about a ring's width. The pixels are corrected with the DataSet's e2dx/e2dy or detector file, as ImageD11 corrects its peaks (the dataset cell says which); if it names none, the widths include the distortion.
- **A ring that wobbles with η in the cake** (a sinusoid): the geometry (beam centre, tilts) is off. Fix the calibration rather than widening `tth_tol`, which would also take in more of the neighbours.
- **Weak rings** add predictions but little intensity: each ring costs time in every step. Six to eight rings are usually enough; more help small grains and lower symmetry. Leaving out a weak ring between strong ones is fine.

```python
rings_given = list(range(rings)) if isinstance(rings, (int, np.integer)) else sorted(int(r) for r in rings)
cake = dg.ring_profile_cake(dset, max(max_rings_shown, rings_given[-1] + 1), cif, merge=ring_merge)
fig = figure("rings", figsize=(14, 9))
table = cake["rings"]
ring_list = widgets.SelectMultiple(
    options=[(f"{r}: {cake['hkl'][r]} {table['tth'][r]:.3f}°", r) for r in range(len(table["tth"]))],
    value=tuple(rings_given), rows=min(len(table["tth"]), 12), description="rings")
w_rings = interact(fig, pl.plot_rings(fig, cake), rings=ring_list,
                   tth_tol=slider(tth_tol or 0.0, 0.0, max(0.3, 2 * (tth_tol or 0.0)), 0.002, "tth_tol"))
```

## 2. The data: `--monitor`, `--gridstep`, `--detector-mask`

The histogram stage puts all the pixels once into two histograms: the lit map (0.5° in η, 0.25° in ω, all rows summed) for pruning, and the fit's data (1° × 1°, per dty row). This is the one pass over the whole dataset (re-run it after changing the rings).

**Detector coverage.** The stage also records the share of each (ring, η bin) that the detector measures: 0 off its edges and in masked pixels (module gaps, dead pixels, the beamstop). It reads the DataSet's segmenter mask, or `detector_mask` in the parameters (ImageD11's convention: non-zero for a masked pixel). The search leaves out predictions on less than half-measured bins, and the fits scale each prediction by its bin's share, so a reflection on a module gap is not taken as a missing grain. The first plot below shows it: dips are gaps and the mask, a ring falling to 0 over a range of η runs off the detector. If a dip lies where you know there is no gap, the mask is wrong; if gaps you know of do not show, the mask is missing.

Then the total intensity of each dty row and of each ω bin. Across dty, a sample gives a smooth profile; **steps from one row to the next** mean that the flux changed between the rows' scans (e.g. a refill), which makes ring artefacts centred on the rotation axis: normalise with `--monitor` (e.g. `fpico6`). Along ω, a slow drift is beam decay within a rotation. Sharp spikes in either are single strong spots, not flux.

`--gridstep` sums rows in groups to make voxels larger: fewer voxels and rows, a much cheaper fit, coarser maps.

Each frame's ω and dty come from the DataSet (`ds.omega`, `ds.dty`, reshaped by ImageD11 to the sinogram's rows), not from the sparse file's raw readings. The last plot is how the DataSet's dty moves within each row. In a fly scan over ω and dty together, the readings drift slowly through each rotation; binned by them, each frame falls in the row of the moment, so the rows alias: gaps in some rows of the sinogram and doubled intensity in others, and a scalloped sample edge. ImageD11 only reshapes the readings: if they drift here, replace `ds.dty` by its mean over each row in ImageD11, save the DataSet, and re-run from the top. The motion within each row is then still followed by the model (`frame_dty`, on by default: each frame's dty is its row's plus the motion read from the sparse file), so each voxel goes into the rows that really saw it. `dty_source = "frames"` instead bins each frame by its own dty, which spreads a helical rotation over 2-3 rows.

```python
rings_used = sorted(w_rings.kwargs["rings"])
if not rings_used:
    raise ValueError("select at least one ring in section 1")
rings_used = len(rings_used) if rings_used == list(range(len(rings_used))) else rings_used  # the first n: a count
histogram_file = ix.stages.stage_histogram(analysisroot, sample, dataset, f"{stem}_histogram.h5", phase=phase, parfile=parfile,
                                    rings=rings_used, ring_merge=ring_merge, cif=cif, tth_tol=w_rings.kwargs["tth_tol"] or None,
                                    monitor=monitor, dty_source=dty_source, frame_dty=frame_dty, gridstep=gridstep, grid_pad=grid_pad, saturation=saturation, pol_factor=pol_factor,
                                    detector_mask=detector_mask, command=NOTEBOOK, log=log)
d = ix.stages.load_histogram(histogram_file)
if d["rings"].get("coverage") is not None:
    fig = figure("coverage", figsize=(15, 3.2))
    pl.plot_coverage(fig, d["rings"]["coverage"], d["rings"]["index"])
    show(fig)
scan = d["scan"]
dty_within = None if dset["ds"]["dty"] is None else np.asarray(dset["ds"]["dty"], float)
fig = figure("rows", figsize=(15, 3.6))
pl.plot_rows(fig, dg.row_profile(d["H"], d["n_rings"], scan), dty_within, scan["ystep"] / gridstep)
show(fig)
```

## 3. Rotation axis: `--y0`

y0 is the dty at which the rotation axis is in the beam. A voxel at (x, y) in the sample is in the beam at dty = y0 − (x sin ω + y cos ω), so any point fixed in the sample traces y0 + b sin ω + c cos ω in the sinogram, and a fit of that curve gives y0. ImageD11's `tomo_2_map` fits each grain's sinogram this way.

Here the point is the middle of the sample's projection: at each ω, halfway between the first and last dty rows above 5% of that projection's maximum (the sample's edges, interpolated between rows). The edges are sharp whichever grains diffract, so this is much steadier than the intensity-weighted mean dty (also shown), which jumps as grains light up and go dark: on the phantom (true y0 0.3, 1 µm steps) the edges give 0.19, the mean −0.03. Both need the whole sample within the scanned dty range.

Check it on the reconstruction, made as ImageD11's `tomo_2_map` makes its whole-sample mask: a sinogram of log(1 + counts) of each (ring, η) bin, so that every spot counts about the same however bright, and filtered back-projection with a Hamming-windowed ramp. With the wrong y0, the sample's edge blurs and the network of grain boundaries inside it doubles and smears; on the phantom that shows clearly at 3 dty steps, faintly at 1. The slider moves y0 by up to 10 steps either way; the reconstruction is on the indexer's voxel grid, with its model of where a voxel is in the beam. y0 is an option of the fit stage.

```python
edges = dg.y0_from_edges(d["H"], d["n_rings"], scan)
y0_dataset = d["prov"]["resolved"].get("y0_dataset")
y0_start = y0_dataset if y0 is None else y0
if y0_start is None:
    raise ValueError("the DataSet has no y0: set y0 in the parameters (this section then helps to check it)")
nr_start = d["grid"]["n"]  # the grid the histogram stage fixed: a y0 moves no voxel
fig = figure("y0", figsize=(15, 4.6))
lo, hi = min(y0_start, edges["y0"]) - 10 * scan["ystep"], max(y0_start, edges["y0"]) + 10 * scan["ystep"]
w_y0 = interact(fig, pl.plot_y0(fig, edges, lambda y: dg.back_projection(d, y, nr_start), y0_start, scan),
                y0=slider(y0_start, lo, hi, scan["ystep"] / 20, "y0"))
```

```python
Y0 = w_y0.kwargs["y0"]  # the rotation axis used from here on
NR = d["grid"]["n"]  # fixed by the histogram stage (with --grid-pad); y0 only says which rows see each voxel
print(f"y0 = {Y0:.4f}; voxels {NR} x {NR} of {d['grid']['step']:g}, centred on the rotation axis")
```

### The sample mask: `--mask`

The fit only needs the voxels the sample fills, and only the voxels the data constrain. The voxel grid is square, but a voxel further from the rotation axis than the nearer end of the scanned dty range is in the beam at some ω only (the grid's corners, and more when y0 is off the middle of the scan). The data hardly constrain those voxels, and MLEM can give them large occupancies that explain almost nothing, which then swamp the `--occupied` cut. So by default (`mask = "disc"`) the fit takes only the voxels seen at every ω: the disc outlined in cyan. A sample a little larger than the disc should still be fitted whole, or its outer part piles its intensity onto the disc's edge: grow the disc with `pad` (a voxel at radius r is in the beam over (2/π) asin(r_disc / r) of the rotation, 2/3 at 1.15 r_disc). A sample reaching past the voxel grid itself needs `grid_pad` in the histogram stage, which fixes the grid for every later stage.

Within the disc, a mask can also follow the sample, as ImageD11's `tomo_2_map` makes its whole-sample mask: the pixels above a threshold (Otsu's by default; the slider moves it), their largest connected region, and its convex hull. Or draw it by hand (`draw = True` in the parameters): click the sample's outline on the image that appears, close it on its first corner, then run the next cell (`anri.index.mask.draw_mask`, as ImageD11's `InteractiveMask`). Then the air around the sample is not fitted either (fewer voxels, no noise orientations there), and the `--occupied` cut of section 10 works within the mask. `mask = "threshold"` in the parameters uses the threshold at the slider; it, or a drawn mask, is saved with its voxel size (`anri.index.mask.save_mask`, an `.npz`) for `--mask`. `mask = "auto"` instead lets the fit stage find the sample's support in the projections (`anri.index.mask.support_mask`: the voxels whose row is lit at every ω, wherever the beam reaches; it rounds sharp corners of a coarse-grained sample). A sample that fills the scan leaves no air to threshold against: keep the disc.

```python
rec = dg.back_projection(d, Y0)
disc = ix.mask.disc_mask(scan, Y0, NR, pad)
fig = figure("mask", figsize=(14, 4.4))
w_mask = interact(fig, pl.plot_mask(fig, rec, disc), threshold=slider(ix.mask.otsu(rec), float(np.percentile(rec, 1)),
                                                                float(np.percentile(rec, 99.9)), float(np.ptp(rec)) / 500,
                                                                "threshold"))
if draw:
    if not interactive:
        raise ValueError("drawing the mask needs interactive = True")
    get_mask = ix.mask.draw_mask(rec)  # draw the outline on this figure, then run the next cell
```

```python
# the fit's --mask: "disc", "auto", a file, or drawn or thresholded at the slider (saved with its voxel size, an .npz)
mask_used = mask
threshold = w_mask.kwargs["threshold"]
if draw or mask == "threshold":
    mask_used = f"{stem}_mask.npz"
    ix.mask.save_mask(mask_used, get_mask() if draw else ix.mask.threshold_mask(rec, threshold) & disc, d["grid"]["step"])
print(f"--mask {mask_used}")
```

## 4. Lit map: `--lit` and `--etacut`

The lit map says where there is intensity: a bin is lit above `lit` × the median non-empty bin. Completeness, the fraction of an orientation's predicted spots that land on lit bins, is the first filter on the orientation grid, so the lit map should be **sparse**: the more of it is lit, the more wrong orientations match by chance (section 5). But a high cut drops weak spots, i.e. small grains. The left plot is the trade-off: the lit area against the intensity it holds, as the cut rises. A good cut is near the elbow, where the area falls fast and the intensity slowly.

```python
fig = figure("lit", figsize=(15, 4.6))
w_lit = interact(fig, pl.plot_lit(fig, dg.lit_curve(d["H_lit"], d["n_rings"], scan), scan, etacut),
                 lit=slider(lit, -1, 2, 0.05, "lit", log_scale=True),
                 ring=widgets.IntSlider(value=0, min=0, max=d["n_rings"] - 1, description="ring", continuous_update=False))
```

Reflections near η = 0 and 180° (12 and 6 o'clock) are poorly placed in ω: a small rotation δ of the crystal moves a spot by up to δ / |sin η| in ω, so near the rotation axis spots smear over many frames and the grid's tolerances blow up. Reflections with |sin η| ≤ `etacut` are not used anywhere (dotted lines above). Raising it costs reflections (2 arcsin(etacut) / π of every ring) and the intensity there.

```python
fig = figure("etacut", figsize=(12, 3.6))
w_eta = interact(fig, pl.plot_etacut(fig, dg.eta_profile(d["H_lit"], d["n_rings"], scan)),
                 etacut=slider(etacut, 0.0, 0.7, 0.01, "etacut"))
```

## 5. Orientation grid: `--grid`, `--max-chance`

The grid covers one fundamental zone. Its step sets how far the truth can be from the nearest grid point, so the matching tolerances, and with them the completeness that wrong orientations reach by chance: the **chance completeness**, the median over a sample of the grid (most grid orientations are wrong). The search stage takes the coarsest step whose chance completeness is at most `max_chance`. A finer grid separates real orientations from decoys better and gives more precise orientations, but has many more orientations (the cost of the search grows with it).

The samples are drawn as `anri.index.search.choose_grid` draws them, so the values match the command line's log. The lit map and η cut chosen above are used; re-run this cell after changing them.

```python
lit_used, etacut_used = w_lit.kwargs["lit"], w_eta.kwargs["etacut"]
lit_map = ix.search.lit_map(d["H_lit"], scan, d["n_rings"], lit_used, etacut_used)
chances = dg.grid_chances(d, lit_map, log=log)
fig = figure("grid", figsize=(7, 3.6))
w_grid = interact(fig, pl.plot_grid(fig, chances, grid), max_chance=slider(max_chance, 0.05, 0.95, 0.05, "max_chance"))
```

**How fine a grid the fit needs.** Chance completeness says how coarse a grid the pruning can work with. The fit itself predicts point spots from grid orientations into 1° × 1° bins: a grain between grid points has its spots up to the grid's worst case away, and a grid much coarser than the bins cannot represent it (MLEM then spreads it over several grid points, differently from voxel to voxel, and grain boundaries go noisy).

Below, for each grid step, the orientations above chance completeness are fitted to the row-summed data (the orientation fit of section 6). The deviance falls as the grid gets finer while the grid limits the fit, and flattens where the data no longer tell finer grids apart. Choose the coarsest step on the flat part (`grid` in the parameters; 0.75° is shown as a check that 1° is fine enough). Set `grid_scan = False` to skip this on big datasets: it costs about one orientation fit per step.

```python
if grid_scan:
    deviances = dg.grid_deviances(d, lit_map, (*chances, 0.75), etacut_used, min_lr, keep, censor, log=log)
    fig = figure("grid_deviances", figsize=(11, 3.6))
    pl.plot_grid_deviances(fig, deviances, min_lr)
    show(fig)
```

## 6. Pruning: `--prune`, `--min-comp`, `--min-lr`, `--keep`

The search stage scores every orientation of the grid by its completeness. Real orientations sit well above the chance level; the bulk of the grid sits at it. The filter stage then chooses the orientations for the fit:

- `--prune completeness` keeps everything at or above `min_comp` (default halfway from chance to the maximum), the most complete first, at most `keep`.
- `--prune likelihood` (default) fits one occupancy per orientation above `min_comp` (default: the chance level) to the row-summed data, and keeps those whose **likelihood ratio** is above `min_lr`: how much the fit worsens without that orientation (about χ² with one degree of freedom for one that is not there, so 25 is about 5σ). Raising `min_comp` here drops decoys, but also small grains, whose completeness is lower.

```python
search_file = ix.stages.stage_search(histogram_file, grid=grid, max_chance=w_grid.kwargs["max_chance"], lit=lit_used,
                              etacut=etacut_used, command=NOTEBOOK, log=log)
search = anri.io.provenance.read_provenance(search_file)
chance, halfway = search["results"]["chance_completeness"], search["results"]["completeness_halfway"]
step = search["resolved"]["grid_step"]
with h5py.File(search_file, "r") as h:
    U_grid, comp = h["U_grid"][()], h["completeness"][()]


def default_cut(prune):
    return chance if prune == "likelihood" else halfway


fig = figure("completeness", figsize=(13, 3.6))
w_comp = interact(fig, pl.plot_completeness(fig, comp, chance, halfway, keep),
                  prune=widgets.Dropdown(options=["likelihood", "completeness"], value=prune, description="prune"),
                  min_comp=slider(default_cut(prune) if min_comp is None else min_comp, 0.0, 1.0, 0.01, "min_comp"))
if interactive:  # a new prune mode starts from its own default min_comp
    w_comp.children[0].observe(lambda change: setattr(w_comp.children[1], "value", default_cut(change["new"])),
                               names="value")
```

The filter stage. With `--prune likelihood`, every orientation above `min_comp` gets one occupancy, fitted by MLEM to the histogram with its dty rows summed (it explains overlapping spots jointly, with Lorentz, polarisation and |F|²), then its likelihood ratio; the file keeps every ratio, so the plot below can redo the choice for any `min_lr` and `keep`.

In the plots: real orientations reach high likelihood ratios. On the right, each orientation's likelihood ratio against its misorientation to the nearest one with a higher ratio. A grain shows as one strong point far from anything stronger (top right: a peak), with a halo of weaker points one or two grid steps from it, its neighbours on the grid soaking up what the grid cannot fit exactly. The number of kept peaks (beyond the population radius, 1.8 grid steps) is roughly the number of distinct orientations: grains and twins. Weak peaks far from everything are small grains or decoys; `min_lr` decides which are kept.

```python
prune_used, min_comp_used = w_comp.kwargs["prune"], w_comp.kwargs["min_comp"]
if min_comp is None and np.isclose(min_comp_used, default_cut(prune_used)):
    min_comp_used = None  # the stage's own default
filter_options = {"prune": prune_used, "min_comp": min_comp_used, "keep": keep, "censor": censor}
filter_file = ix.stages.stage_filter(search_file, min_lr=min_lr, **filter_options, command=NOTEBOOK, log=log)
if prune_used == "likelihood":
    with h5py.File(filter_file, "r") as h:
        fitted, lr = h["fitted"][()], h["likelihood_ratio"][()]
    fig = figure("likelihood", figsize=(16, 3.9))
    w_lr = interact(fig, pl.plot_likelihood(fig, fitted, lr, comp[fitted], dg.nearest_stronger(U_grid[fitted], lr, d["ops"]),
                                            step, anri.io.provenance.read_provenance(filter_file)["resolved"]["min_comp_used"]),
                    min_lr=slider(min_lr, 0, 4, 0.05, "min_lr", log_scale=True),
                    keep=widgets.BoundedIntText(value=keep, min=1, max=10**8, description="keep"))
```

```python
# min_lr and keep moved on the sliders: the filter again with them (its orientation fit too), so that its file says so
if prune_used == "likelihood" and (w_lr.kwargs["min_lr"], w_lr.kwargs["keep"]) != (min_lr, keep):
    filter_options["keep"] = w_lr.kwargs["keep"]
    filter_file = ix.stages.stage_filter(search_file, min_lr=w_lr.kwargs["min_lr"], **filter_options, command=NOTEBOOK, log=log)
with h5py.File(filter_file, "r") as h:
    print(f"{len(h['kept'])} orientations kept for the occupancy fit")
```

## 7. Occupancy: `--cand`, `--iter`, `--coarse`, `--block-gb`, `--beam`, `--censor`

The fit stage. Each voxel keeps its `cand` best orientations (by the first MLEM update from flat occupancy), and MLEM fits their occupancies, every voxel jointly. It costs the same as on the command line; time grows with voxels × candidates × reflections × iterations. A callback records the deviance at every iteration and the map at a few.

- `coarse`: first fit voxels this many times larger, then give each voxel the candidates of its coarse neighbourhood (~16× cheaper at 4 on large maps).
- `block_gb` only sets how many voxels are handled at once (memory, not results).
- `beam`: the beam's FWHM across dty. Each voxel is spread over the dty rows by the beam's profile (a Gaussian of that FWHM) integrated over the voxel at each ω, so a beam wider than the step (overfocusing) puts a voxel into more rows. With FWHM = step it is close to the default, a linear spread over the 2 nearest rows (`beam = 0`), and maps come out as good (AP1_1, by the figure of merit), at ~2.4× the cost of the candidate pass.
- `censor`: ImageD11 keeps only pixels above its segmentation cut, so an empty bin means every pixel in it was below the cut, not zero. Where the model predicts fewer than `censor` counts in an empty bin, the bin counts as agreeing with the model. That stops weak orientations, whose faint spots fell below the cut, from being driven to zero. Roughly the cut times the pixel-frames a faint spot covers; 0 turns it off.

```python
if run_fit:
    callback, record = dg.convergence_recorder(n_iter, log=log)
    fit_file = ix.stages.stage_fit(filter_file, y0=None if y0_dataset is not None and np.isclose(Y0, y0_dataset) else Y0,
                            cand=n_cand, n_iter=n_iter, coarse=coarse, beam=beam, mask=mask_used, pad=pad, absorption=absorption, mu=mu, block_gb=block_gb,
                            censor=censor, save_model=True, command=NOTEBOOK, log=log, callback=callback)
    with h5py.File(fit_file, "r") as h:
        f, cand, U_kept, fitted_v = h["f"][()], h["cand"][()], h["U"][()], h["mask"][()]
        model, row_ratio = h["model"][()], h["row_ratio"][()]
    tot = f.sum(1)
```

**Iterations.** Each iteration is a SQUAREM step: three MLEM updates and an extrapolation, which goes about as far as ten or more plain updates. The deviance flattens early, but thin features (twins one or two voxels thick, small grains) keep changing long after. More iterations help them, while fewer keep each population's mean a little closer to the truth between grid points (a fully converged fit puts each grain's occupancy on fewer grid points): on the def316l phantom, 5 iterations gave the best mean orientation, 10 or more the best twins. The middle plots show how much of each occupied voxel's occupancy moved between orientations from one snapshot to the next (half the summed change of its occupancies, as a share of its total, per iteration), and where it still moves at the end. If that is still falling steeply at the last iteration, more iterations will change the map.

**Candidates.** Each voxel's occupancy, sorted, as a share of the voxel's total: if the last candidates still hold a noticeable share in many voxels, the voxel wanted more than `cand` orientations.

```python
if run_fit:
    occupied_v = ix.populations.select_occupied(tot, fitted_v, occupied)  # at the parameters' --occupied; section 10 chooses it
    fig = figure("convergence", figsize=(18, 4))
    pl.plot_convergence(fig, record, f, fitted_v, occupied_v, NR)
    show(fig)
    tail = dg.candidate_shares(f, occupied_v)[:, -max(1, n_cand // 8) :].sum(1)
    print(f"the last {max(1, n_cand // 8)} candidates hold {np.median(tail) * 100:.2f}% of an occupied voxel's "
          f"occupancy (median), {np.percentile(tail, 99) * 100:.2f}% (99th percentile)")
```

## 8. Model against data

The fitted histogram against the data.

- **Per dty row** (as the fit stage logs it): measured / fitted intensity. Steps from row to row point to the flux varying between the rows' scans (`--monitor`); a smooth trend with radius, to the model.
- **Per ring and spot by spot**: the data, the model and the difference for one ring with every row summed (η against ω), and the sinogram of an η window. Compact positive spots in the difference are intensity that no kept orientation explains: a missed grain (lower `min_lr`, `lit` or `min_comp`, or a finer grid) or another phase. Negative spots are predicted where there is nothing: decoys, or wrong orientations. A smooth residual all along a ring is background, which the model does not have.

Only η bins the fit models are used in the ratios (|sin η| above `etacut`).

```python
if run_fit:
    fig = figure("ratios", figsize=(13, 3.6))
    pl.plot_row_ratios(fig, scan["dty0"] + np.arange(scan["n_rows"]) * scan["ystep"], row_ratio,
                       dg.ring_ratios(d["H"], model, d["n_rings"], scan, etacut_used))
    show(fig)
    fig = figure("residual", figsize=(16, 8))
    w_res = interact(fig, pl.plot_residual(fig, d["H"], model, d["n_rings"], scan),
                     ring=widgets.IntSlider(value=0, min=0, max=d["n_rings"] - 1, description="ring",
                                            continuous_update=False),
                     eta=widgets.IntRangeSlider(value=(80, 100), min=-180, max=180, step=1, description="η window",
                                                continuous_update=False))
```

## 9. Local search: `--levels`, `--cover`, `--per-voxel`

The fit's orientations are points of the search grid, up to 0.87 grid steps from the truth. Where peaks are sharper than that, the fit shares a grain's peaks among grid neighbours, and a twin whose grid point lands nearer the exact twin than the grain's own lands to the grain takes the reflections the two share: a ghost twin, where the sample has none. The local search stage replaces every orientation a voxel uses by the 27 rotations of a 3 × 3 × 3 grid around it, a third of the step apart, and fits each voxel again over its own orientations' children only, with the fit's options (`anri.index.local`). Each level costs about one fit.

- `levels`: each a third of the step before (2: a 1° grid to 0.11°). Past the fit's 1° bins the deviance barely falls, but thin lamellae still sharpen. 0 skips the stage.
- `cover`, `per_voxel`: a voxel uses its most occupied orientations that together hold `cover` of its occupancy, at most `per_voxel`. Each becomes 27 candidates, so a level fits up to 27 × `per_voxel` per voxel.

**Before running it.** How many of the fit's orientations each occupied voxel needs to hold `cover` of its occupancy (`anri.index.diagnostics.orientations_needed`), and where. Voxels needing more than `per_voxel` (red on the map) lose the rest of their occupancy to the orientations they keep. A few, at grain boundaries and in lamellae, are expected; red over whole grains means the fit spread each grain over many grid points: raise `per_voxel`, or lower `cover`. The summary gives the number of orientations used and the children a level predicts, which sets its memory and time.

```python
if run_fit and levels > 0:
    fig = figure("local_slots", figsize=(14, 4.2))
    w_slots = interact(fig, pl.plot_local_slots(fig, f, cand, fitted_v, occupied_v, NR),
                       cover=slider(cover, 0.5, 1.0, 0.01, "cover"),
                       per_voxel=widgets.IntSlider(value=per_voxel, min=1, max=f.shape[1], description="per_voxel",
                                                   continuous_update=False))
```

The local search, with `cover` and `per_voxel` from the sliders. A callback records the deviance at every iteration of every level (`anri.index.diagnostics.level_recorder`). Then each occupied voxel's main population (grouped as in section 10, within 1.8 grid steps) before and after (`anri.index.diagnostics.local_change`):

- **Deviance**: the fit's iterations, then each level's. A level starts from each parent's occupancy shared among its children, then falls below where the fit ended. Little gain at the last level means more levels will not change the fit to the histogram (thin features can still sharpen).
- **How far the main population moved**, as a map and a histogram. Most voxels move by up to the grid's worst case (red line): the fit's orientations were that far off. Voxels moving by about a twin's misorientation (60° for Σ3) changed their main population: a ghost twin that held the voxel on the grid has lost it, or a thin real twin has won its voxels from the grain. Check that such voxels follow lamellae or boundaries, not noise.
- **Main population fraction and spread**: with ghosts gone, more of a voxel's occupancy falls in its main population, and its spread shrinks with the finer grid (it still includes the last level's spacing).

```python
if run_fit and levels > 0:
    level_callback, level_record = dg.level_recorder()
    local_file = ix.stages.stage_localsearch(fit_file, levels=levels, cover=w_slots.kwargs["cover"],
                                             per_voxel=w_slots.kwargs["per_voxel"], command=NOTEBOOK, log=log,
                                             callback=level_callback)
    change = dg.local_change(fit_file, local_file, d["ops"], 1.8 * step, occupied_v)
    fig = figure("local", figsize=(18, 4))
    pl.plot_local(fig, record["deviance"], level_record["deviance"], change,
                  search["resolved"]["grid_worst_case_deg"], NR)
    show(fig)
    with h5py.File(local_file, "r") as h:
        f, cand, U_kept = h["f"][()], h["cand"][()], h["U"][()]
    tot = f.sum(1)
else:
    local_file = fit_file if run_fit else None
```

## 10. Populations: `--occupied`, `--min-frac`

Each voxel's occupied candidates are grouped into populations (candidates within 1.8 grid steps of the most occupied one join it, and so on), each with a fraction of the voxel, a mean orientation and a spread (which includes the grid's own spacing, so it is an upper bound).

- `occupied`: voxels in the mask (section 3) count as sample above this × the 99th percentile of the total occupancy in the mask; voxels outside it are never occupied. The histogram of the total occupancy should show the sample well apart from the air around it; put the cut in the gap. (The raw occupancy is always saved.)
- `min_frac`: populations holding at least this fraction of a voxel are reported. Second populations are real where two grains (or a grain and its twin) share a voxel, at boundaries and in lamellae; a scatter of weak ones all over the grains is noise from the coarse model (the beam is wider than a voxel, the grid coarser than the truth). Put the cut above that floor.

```python
if run_fit:
    frac, _, spread, _ = ix.populations.group(f, cand, U_kept, d["ops"], 1.8 * step)
    fig = figure("populations", figsize=(15, 9))
    w_pop = interact(fig, pl.plot_populations(fig, tot, fitted_v, frac, spread, step, NR),
                     occupied=slider(occupied, 0.0, 1.0, 0.01, "occupied"),
                     min_frac=slider(min_frac, 0.0, 0.5, 0.01, "min_frac"))
```

The populations stage writes the populations, every population as map entries for the renderer (`_entries.npz`) and, with ImageD11, a TensorMap of the main population (`_tmap.h5`, and `.xdmf` for ParaView). Below, the main population's IPF colours (with ImageD11 and orix installed), dimmed by each voxel's occupancy (`ipf_z_occupancy`: full brightness at the 99th percentile of the occupied voxels, as ImageD11 weights its IPF maps by intensity, so thin lamellae and weakly diffracting regions show), and its completeness: the fraction of its predicted reflections that land on lit bins. Low completeness inside the sample points to a wrong orientation, or a voxel the data barely see.

```python
if run_fit:
    populations_file = ix.stages.stage_populations(local_file, occupied=w_pop.kwargs["occupied"],
                                            min_frac=w_pop.kwargs["min_frac"], command=NOTEBOOK, log=log)
    with h5py.File(populations_file, "r") as h:
        occ, comp_main = h["occupied"][()], h["comp_pop"][:, 0]
    try:
        from ImageD11.sinograms.tensor_map import TensorMap

        tmap = TensorMap.from_h5(populations_file[: -len(".h5")] + "_tmap.h5")
        ipf = np.where((tmap.phase_ids[0] >= 0)[..., None], tmap["ipf_z_occupancy"][0], 1.0)
    except (ImportError, OSError, KeyError):  # KeyError: no IPF maps (orix not installed)
        ipf = None
    fig = figure("main", figsize=(11, 5))
    pl.plot_main_population(fig, np.where(occ, comp_main, np.nan), ipf, NR)
    show(fig)
```
